## 1. Import Library dan Load Dataset Iris

Pada praktikum ini digunakan dataset Iris untuk mencoba algoritma Naïve Bayes. Pada tahap awal, library yang diperlukan diimport terlebih dahulu, kemudian dataset Iris diambil dari Scikit-Learn.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

## 2. Mengubah Dataset Menjadi DataFrame

Dataset Iris yang sudah diambil kemudian diubah ke dalam bentuk DataFrame menggunakan Pandas. Setelah itu ditambahkan kolom Species sebagai target atau label dari setiap data.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

## 3. Melihat Data Target

Variabel `y` digunakan untuk melihat target atau label dari dataset Iris. Data target terdiri dari tiga spesies, yaitu setosa, versicolor, dan virginica.

```python
y
```

## 4. Mengubah Label Menjadi Data Numerik

Label spesies yang masih berbentuk teks diubah menjadi angka menggunakan `LabelEncoder`. Dengan cara ini, setiap kelas akan mempunyai nilai numerik yang bisa digunakan dalam proses klasifikasi.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

## 5. Membagi Data Training dan Testing

Data Iris kemudian dibagi menjadi data training dan data testing. Data training digunakan untuk melatih model, sedangkan data testing digunakan untuk menguji hasil prediksi model.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

## 6. Menentukan Gaussian Naïve Bayes

Dataset Iris memiliki fitur berupa data numerik kontinu, seperti panjang dan lebar sepal serta petal. Karena jenis datanya berupa angka kontinu, pada praktikum ini digunakan Gaussian Naïve Bayes.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

## 7. Melakukan Prediksi Data Testing

Setelah model selesai dilatih, model digunakan untuk melakukan prediksi terhadap data testing. Hasil prediksi disimpan pada variabel `result`.

```python
result = NB_model.predict(X_test)
result
```

## 8. Menghitung Accuracy

Setelah mendapatkan hasil prediksi, dilakukan perhitungan accuracy untuk melihat seberapa banyak data testing yang berhasil diprediksi dengan benar oleh model.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

## 9. Melakukan Prediksi untuk Data Baru

Selain data testing, model juga dapat digunakan untuk memprediksi data baru. Pada tahap ini diberikan satu input baru yang memiliki empat nilai fitur seperti pada dataset Iris.

```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

## 10. Mengatasi Warning pada Prediksi Data Baru

Pada percobaan sebelumnya muncul warning karena model dilatih menggunakan DataFrame yang memiliki nama kolom, sedangkan input baru diberikan dalam bentuk list atau array tanpa nama kolom. Supaya format input sama dengan data saat training, input baru dibuat menjadi DataFrame dengan nama kolom yang sama.

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Pada bagian ini dilakukan prediksi terhadap satu data baru yang diberikan pada soal. Model Gaussian Naïve Bayes yang sudah dibuat sebelumnya digunakan untuk menentukan spesies dari data tersebut.
